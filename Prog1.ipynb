```python
import pandas as vp
data = vp.read_csv("data2.csv")

data
```

```python
x = data.iloc[:, :-1]
y = data.iloc[:, -1]
y
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
x["CGPA"] = x["CGPA"].map({"Low": 0, "Medium": 1, "High": 2})

x["Communication"] = x["Communication"].map({"Bad": 0, "Good": 1})

x["Aptitude"] = x["Aptitude"].map({"Low": 0, "Medium": 1, "High": 2})

x["Programming_skill"] = x["Programming_skill"].map({"Bad": 0, "Good": 1})



print(x)
```

```python
from sklearn.tree import DecisionTreeClassifier
sk = DecisionTreeClassifier()
sk.fit(x, y)
print(sk.predict([[3, 3, 1, 1]]))
sk.predict([[1, 2, 1, 2]])
```
