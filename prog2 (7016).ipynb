```python
import pandas as vp
data = vp.read_csv("data1.csv")

data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 5))

plt.plot(data.index, data["average-mileage"], marker="o")

plt.title("Average Mileage of Cars")
plt.xlabel("Car Index")
plt.ylabel("Average Mileage")
plt.grid(True)

plt.show()
```

```python
company_count = data["company"].value_counts()

plt.figure(figsize=(12, 6))

company_count.plot(kind="bar")

plt.title("Number of Cars by Company")
plt.xlabel("Company")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(data["horsepower"], data["price"])

plt.title("Horsepower vs Car Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(data["price"].dropna(), bins=10, edgecolor="black")

plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```
